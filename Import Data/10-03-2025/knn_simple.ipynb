# K-Nearest Neighbors (K-NN)


## Importar las librerías


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importar el conjunto de datos


```python
datos = pd.read_csv("Social_Network_Ads.csv")
X = datos.iloc[:, :-1].values
y = datos.iloc[:, -1].values
```

## División del conjunto de datos en un conjunto para entrenar y otro para probar


```python
from sklearn.model_selection import train_test_split



X_entreno, X_prueba, y_entreno, y_prueba = train_test_split(
    X, y, test_size=0.25, random_state=0
)
```

```python
print(X_entreno)
```

```python
print(y_entreno)
```

```python
print(X_prueba)
```

```python
print(y_prueba)
```

## Escalamiento o Normalización de variables


```python
from sklearn.preprocessing import StandardScaler



normalizador = StandardScaler()


X_entreno = normalizador.fit_transform(X_entreno)


X_prueba = normalizador.transform(X_prueba)
```

```python
print(X_entreno)
```

```python
print(X_prueba)
```

## Entrenar el modelo K-NN con el conjunto de datos para entrenamiento


```python
from sklearn.neighbors import KNeighborsClassifier

clasificador = KNeighborsClassifier(n_neighbors=5)
clasificador.fit(X_entreno, y_entreno)
```

## Predecir un nuevo resultado


```python
print(clasificador.predict(normalizador.transform([[30, 87000]])))
```

## Predecir los datos del conjunto de pruebas


```python
y_pred = clasificador.predict(X_prueba)
print(
    np.concatenate(
        (y_pred.reshape(len(y_pred), 1), y_prueba.reshape(len(y_prueba), 1)), 1
    )
)
```

## Obtener la Matriz de Confusión


```python
from sklearn.metrics import confusion_matrix, accuracy_score



mat_conf = confusion_matrix(y_prueba, y_pred)


print(mat_conf)


accuracy_score(y_prueba, y_pred)
```

## Visualizar los resultados del conjunto de datos para entrenamiento


```python
from matplotlib.colors import ListedColormap



X_set, y_set = normalizador.inverse_transform(X_entreno), y_entreno


X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 10, stop=X_set[:, 0].max() + 10, step=1),
    np.arange(start=X_set[:, 1].min() - 1000, stop=X_set[:, 1].max() + 1000, step=1),
)


plt.contourf(
    X1,
    X2,
    clasificador.predict(
        normalizador.transform(np.array([X1.ravel(), X2.ravel()]).T)
    ).reshape(X1.shape),
    #             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
    alpha=0.75,
    cmap=ListedColormap(["red", "green"]),
)
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    #    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)


    plt.scatter(
        X_set[y_set == j, 0],
        X_set[y_set == j, 1],
        color=ListedColormap(["red", "green"])(i),
        label=j,
    )


plt.title("K-NN (Datos para entrenamiento)")


plt.xlabel("Edad")


plt.ylabel("Salario Estimado")


plt.legend()


plt.show()
```

## Visualizar los resultados del conjunto para pruebas


```python
from matplotlib.colors import ListedColormap



X_set, y_set = normalizador.inverse_transform(X_prueba), y_prueba


X1, X2 = np.meshgrid(
    np.arange(start=X_set[:, 0].min() - 10, stop=X_set[:, 0].max() + 10, step=1),
    np.arange(start=X_set[:, 1].min() - 1000, stop=X_set[:, 1].max() + 1000, step=1),
)


plt.contourf(
    X1,
    X2,
    clasificador.predict(
        normalizador.transform(np.array([X1.ravel(), X2.ravel()]).T)
    ).reshape(X1.shape),
    alpha=0.75,
    cmap=ListedColormap(["red", "green"]),
)


plt.xlim(X1.min(), X1.max())


plt.ylim(X2.min(), X2.max())


for i, j in enumerate(np.unique(y_set)):


    plt.scatter(
        X_set[y_set == j, 0],
        X_set[y_set == j, 1],
        color=ListedColormap(["red", "green"])(i),
        label=j,
    )


plt.title("K-NN (Datos de prueba)")


plt.xlabel("Edad")


plt.ylabel("Salario Estimado")


plt.legend()


plt.show()
```
